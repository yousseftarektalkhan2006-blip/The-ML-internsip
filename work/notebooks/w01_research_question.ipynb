## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd
import numpy as np



df = pd.read_csv("The-ML-internsip/data/raw/content_refresh_anonymized.csv")

print(f"Total pages in dataset: {len(df):,}")

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"\nStale + visible pages: {len(stale_visible):,}")

stale_visible_declining = stale_visible[stale_visible["trend_direction"] == "down"]
pct_declining = len(stale_visible_declining) / len(stale_visible) * 100
print(f"Of those, declining: {len(stale_visible_declining):,} ({pct_declining:.1f}%)")

low_ctr_visible = df[
    (df["impressions_90d"] >= 500) &
    (df["avg_position"] > 0) &
    (df["avg_position"] <= 10) &
    (df["ctr"] < 0.5)
]
print(f"\nPage-1 pages with low CTR: {len(low_ctr_visible):,}")
```
